## 🔄 NumPy `sort()` vs Python `sorted()`

#### 1️⃣ NumPy `sort()`
- Return a sorted copy of an array.
- 🧩 **Library function** → Part of NumPy, designed for arrays.  
- ⚡ **Optimized for speed** → Uses C backend, very fast for large numerical data.  
- 📤 **Returns a new sorted array** (original unchanged).  
- 🔄 Can sort along a specific **axis** in multi‑dimensional arrays.  
-  https://numpy.org/doc/stable/reference/generated/numpy.sort.html

#### 2️⃣ Python `sorted()`
- 🧩 **Built‑in function** → Works on any iterable (lists, tuples, sets, strings).  
- 🐍 **General purpose** → Not optimized for numerical computing.  
- 📤 **Returns a new sorted list** (original unchanged).  
- 🎯 Flexible → Can use `key` and `reverse` parameters for custom sorting.  

<br>


| Feature                | NumPy `sort()` 🧩 | Python `sorted()` 🐍 |
|-------------------------|------------------|----------------------|
| **Belongs to**          | NumPy library    | Built‑in function    |
| **Works on**            | NumPy arrays     | Any iterable (list, tuple, set, string) |
| **Return type**         | New NumPy array  | New Python list      |
| **Speed**               | Very fast (C backend, optimized for large arrays) ⚡ | Slower for large data, general purpose |
| **Axis support**        | Yes (can sort along rows/columns in multi‑dim arrays) 🔄 | No axis concept, sorts entire iterable |
| **Custom key support**  | Limited (no `key` parameter) | Flexible (`key` and `reverse` supported) 🎯 |
| **Original data**       | Unchanged (non‑destructive) | Unchanged (non‑destructive) |
| **Best use case**       | Numerical computing, large datasets 📊 | General collections, custom sorting logic 📝 |

  - **One argument** → indices.  
  - **Three arguments** → conditional selection.

## 7️⃣ 🚀 `np.argmax`

#### 🅰️ Purpose
- Returns the **index of the maximum value** in an array.  
- If the array is multi‑dimensional, you can specify an **axis**.  
- `axis=0`=**column‑wise max indices**; `axis=1` → **row‑wise max indices** 
- https://numpy.org/doc/stable/reference/generated/numpy.argmax.html



#### 🅱️ Basic Example
```python
import numpy as np
a = np.array([10, 50, 30, 20])
print(np.argmax(a))   # 1
```
👉 Index `1` because `50` is the largest element.


#### `𝓒.` With Axis
```python
b = np.array([[1, 7, 3],
              [9, 2, 5]])

print(np.argmax(b, axis=0))  # [1 0 1]
print(np.argmax(b, axis=1))  # [1 0]
```

- `axis=0` → column‑wise max indices  
  - Col1 → max at row1 (`9`) → index 1  
  - Col2 → max at row0 (`7`) → index 0  
  - Col3 → max at row1 (`5`) → index 1  

- `axis=1` → row‑wise max indices  
  - Row1 → max at col2 (`7`) → index 1  
  - Row2 → max at col0 (`9`) → index 0  


#### 📝 Exam‑style Takeaway
- ✨ `np.argmax(a)` → index of max in flattened array.  
- ✨ `np.argmax(a, axis=0)` → indices of max values down each column.  
- ✨ `np.argmax(a, axis=1)` → indices of max values across each row.

## **13.** 🚀 `np.corrcoef`

- Return Pearson product-moment correlation coefficients.
- https://numpy.org/doc/stable/reference/generated/numpy.corrcoef.html

#### Parameters
- **`x, y`** → Input arrays.

  - If both `x` and `y` are given → correlation between `x` and `y`.  
- **`rowvar`** → Default `True`.  
  - If `True`, each row represents a variable.  
  - If `False`, each column represents a variable.  
- **`bias`, `ddof`** → Deprecated; ignore in modern usage.


#### 🔹 What It Does
- Computes the **Pearson correlation coefficient matrix**.  
- Formula:  

![Pearson Correlation](../Pandas/Content/1.png)

  - `0` → no correlation

#### ✅ **Takeaway**:  
- `np.corrcoef` gives you the **correlation matrix**.

## **24.** 🚀 `np.allclose`

- **Purpose:** Compares `two arrays element-wise to check `if they are `approximately equal` within a tolerance.
- **Syntex:** `np.allclose(a, b, rtol=1e-05, atol=1e-08)`


| Parameter | Meaning | Default | Example Impact |
| --- | --- | --- | --- |
| `a`, `b` | Arrays to compare | Required | ``[1,2,3] ``vs ``[1,2.000001,3]`` |
| ``rtol`` | Relative tolerance | ``1e-05`` | Allows small % differences |
| ``atol`` | Absolute tolerance | ``1e-08`` | Allows fixed small differences |
| Return | Boolean | — | True/False |

### np.allclose `Numpy` vs `Pandas`

| Feature | ``np.allclose`` (NumPy) | ``equals`` (Pandas) |
| --- | --- | --- |
| **Tolerance** | Yes (``rtol``, ``atol``) | No (strict match) |
| **Data Types** | NumPy arrays | Pandas objects |
| **Use Case** | Floating-point comparisons | Exact equality check |
| **Return** | Boolean (True/False) | Boolean (True/False) |

#### 🍄 **Example A.** `True` Scenario

````python

# 25. np.allclose and equals
import numpy as np

a = np.array([1.0, 2.0, 3.0])
b = np.array([1.0, 2.0000001, 3.0])

print(np.allclose(a, b))  # True (difference within tolerance)
````

#### 🍄 **Example B.** `False` Scenario

````python

a = np.array([1.0, 2.0, 3.0])
b = np.array([1.0, 5.0, 3.0])

print(np.allclose(a, b))  # False (difference is too big)
````
