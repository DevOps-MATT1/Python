# 🌟 Module

A **module** in Python is essentially a file that contains Python code — `functions`, `classes`, and `variables` — that you can import and reuse in other programs. Think of it as a toolbox 🧰 that organizes related functionality together.

### 📘 Components of a Module
- **Functions** → Reusable blocks of code (e.g., `math.sqrt()`).
- **Classes** → Blueprints for creating objects (e.g., `datetime.datetime`).
- **Variables/Constants** → Values defined inside the module (e.g., `pi` in `math`).
- **Executable Code** → Sometimes modules include code that runs when imported.

### ✅ Types of Modules
| **Type**          | **Example** | **Description** |
|--------------------|-------------|-----------------|
| Built‑in Modules   | `math`, `os`, `sys` | Already included with Python. |
| User‑defined       | `my_module.py` | Created by you for custom logic. |
| External Libraries | `numpy`, `pandas` | Installed via `pip`. |

### 📝 Exam‑style Summary
- A **module = Python file** with `.py` extension.  
- It can contain **functions, classes, variables, constants, and code**.  
- Modules help in **code reusability, organization, and maintainability**.  
- Import using `import module_name` or `from module_name import function`.

````python
help('modules')
````

### Method 1: Import Built-In Module

````python
import statistics

num = [1,2,3,4,5]

statistics.mean(num)
````

### Method 2: Import Built-In Module (using Alias)

````python
import statistics as s

num = [1,2,3,4,5]

s.mean(num)
````

### Method 3 : Import Built-In Module (only calling specific function)

````python
from statistics import  mean

num = [1,2,3,4,5]

mean(num)
````

## 2. 🐍 Python `External Modules`:

**Python external modules are third‑party libraries that you install separately (usually via `pip`) to extend Python’s functionality beyond the built‑in modules. They cover areas like data analysis, machine learning, web development, visualization, and more.**  


<br>


## 🔍 What Are External Modules?
- **Definition**: Pre‑written code created by other developers, not included in Python’s standard library.  
- **Installation**: Typically installed using `pip install <module_name>`.  
- **Purpose**: Saves time by providing ready‑made solutions for specialized tasks (e.g., handling data, building web apps, scraping websites).  
- **Source**: Most external modules are hosted on **PyPI (Python Package Index)**.  


<br>


## 📘 Common External Modules (with Use Cases)

| **Category**             | **Popular Modules**                        | **Use Case**                         |
|--------------------------|--------------------------------------------|--------------------------------------|
| **Web Development**      | `Django`, `Flask`                          | Build websites & APIs                |
| **Data Analysis**        | `pandas`, `numpy`                          | Handle datasets, numerical computing |
| **Machine Learning**     | `scikit‑learn`, `TensorFlow`, `PyTorch`    | ML models, deep learning             |
| **Data Visualization**   | `matplotlib`, `seaborn`, `plotly`          | Charts, graphs, dashboards           |
| **Web Scraping**         | `BeautifulSoup`, `Scrapy`                  | Extract data from websites           |
| **Networking/APIs**      | `requests`, `httpx`                        | Make HTTP requests, consume APIs     |
| **Databases**            | `SQLAlchemy`, `psycopg2`, `PyMongo`        | Work with SQL/NoSQL databases        |
| **Testing**              | `pytest`, `unittest`                       | Automated testing                    |
| **GUI Development**      | `PyQt`, `Kivy`, `Tkinter (external forks)` | Build desktop apps                   |
| **Scientific Computing** | `SciPy`, `SymPy`                           | Advanced math, symbolic computation  |

## ⚠️ Risks & Considerations
- **Version conflicts**: External modules may clash with each other; use virtual environments (`venv`, `conda`) to isolate projects.  
- **Security**: Only install from trusted sources (PyPI) to avoid malicious packages.  
- **Maintenance**: Some modules may become outdated or unsupported. Always check documentation and community activity.

#### 1. Using `dir()`
- Returns a list of all names (functions, classes, constants) defined in the module.
- Example output for `math`: `['acos', 'asin', 'atan', 'ceil', 'cos', 'pi', ...]`.

````python
import math
print(dir(math))
````

#### 2. Using `help()`
   
- Opens an interactive help page with detailed documentation.
- Shows descriptions of functions, parameters, and usage.

````python
import math
help(math.acos)
````

#### 3. Using `inspect` Module
   
- Lists all functions in the module with their names and references.
- More precise than dir() because you can filter by type (functions, classes, etc.).

````python
import inspect, math

functions = inspect.getmembers(math, inspect.isbuiltin)
for name, func in functions:
    print(name)


````
